# AI Personal Asistant

```python
# queries answer
# emails => summarize
class PersonalAssistant:
  def __init__(self):
    print("Hi, I am you AI assistant .How can i Help ?")

  def ans_query(self):
    question = input("Ask me anything:")
    response = client.responses.create(
        model = "gpt-6-luna",
        input = [
            {"role" : "system","content": "Act like a helpful personal assistant"},
            {"role": "user", "content":question}
        ],
        max_output_tokens = 512
    )
    print(response.output_text.strip())

  def summarize_email(self):
    email_text = input("paste your email here :")
    prompt = f"summarize the following email in 2-3 sentences :{email_text}"

    response = client.responses.create(
        model = "gpt-6-luna",
        input = [
            {"role": "system","content":"Act like an expert email assitant"},
            {"role":"user" , "content" : prompt}
        ],
        max_output_tokens = 512
    )
    print("\n\nSummary: ", response.output_text.strip())
```

```python
assistant = PersonalAssistant()
```

```python
assistant.ans_query()
```

```python
assistant.summarize_email()
```

# Images in OpenAI

## Responses API

```python
import base64
# images => base64 encoded strings => decode
```

```python
response = client.responses.create(
    model="gpt-6-luna",
    input="Generate an image of a highly detailed futuristic spaceship, flying near a nebula",
    tools=[{"type": "image_generation"}]
)

image_data = [
    output.result
    for output in response.output
    if output.type == "image_generation_call"
]

# decode & store image in a file
if image_data:
    image_base64 = image_data[0]
    with open("spaceship.png", "wb") as f:
        image_bytes = base64.b64decode(image_base64)
        f.write(image_bytes)
```

```python
from IPython.display import display, Image

display(Image(filename="/content/spaceship.png"))
```

#Image API

```python
prompt = "Generate an image of a highly detailed, futuristic spaceship flying near a nebula"

result = client.images.generate(
    model="gpt-image-2.5-sunburst",
    prompt=prompt
)

image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

# Save the image to a file
with open("spaceship-image.png", "wb") as f:
    f.write(image_bytes)
```

```python
display(Image(filename="/content/spaceship-image.png"))
```

```python
# EDIT

prompt = "Take the reference image and convert the spaceship color to neon red"

result = client.images.edit(
    model="gpt-image-2.5-sunburst",
    prompt=prompt,
    image=[
        open("spaceship-image.png", "rb"),
    ]
)

image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

# Save the image to a file
with open("spaceship-red.png", "wb") as f:
    f.write(image_bytes)
```

```python
display(Image(filename="/content/spaceship-red.png"))
```
